```python
from optimum.intel.openvino import OVModelForSpeechSeq2Seq
from transformers import AutoProcessor
import librosa
```

```python
model_id = r"C:\Users\utente\Desktop\chat bot\whisper_large_v3_turbo"
save_dir = "whisper_ov_int8"


# لود و تبدیل مستقیم
# استفاده از quantization_config برای تبدیل به INT8 در حین اکسپورت
model = OVModelForSpeechSeq2Seq.from_pretrained(
    model_id,
    export=True,
    quantization_config={"bits": 8} # اینجا مستقیم میگیم INT8 بشه
)

# ذخیره مدل و پردازشگر
model.save_pretrained(save_dir)
processor = AutoProcessor.from_pretrained(model_id)
processor.save_pretrained(save_dir)

print("Done!")
```

```python
MODEL_PATH = r"C:\Users\utente\Desktop\chat bot\whisper_ov_int8"

processor = AutoProcessor.from_pretrained(MODEL_PATH)
model = OVModelForSpeechSeq2Seq.from_pretrained(MODEL_PATH, device="CPU")

audio, _ = librosa.load(r"C:\Users\utente\Desktop\chat bot\vc\Test 1_16k.wav", sr=16000)

inputs = processor(audio, sampling_rate=16000, return_tensors="pt").input_features
ids = model.generate(inputs)

text = processor.batch_decode(ids, skip_special_tokens=True)[0]
print(text)
```
